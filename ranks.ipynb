```python
import pandas as pd
import math
```

```python
pd.set_option("display.max_rows", None)
positions = ["QB", "RB", "TE", "WR"]
adp = pd.read_csv('ADP.csv')

for position in positions:

    df = pd.read_csv(f'{position}_weekly_stats.csv')
    df["Average"] = df.filter(like="FPTS_WK").mean(axis=1)
    df["StdDev"] = df.filter(like="FPTS_WK").std(axis=1)
    df["ProxySharpe"] = (df["Average"] * 17) / (df["StdDev"] * math.sqrt(17))
    df = df.dropna()
    df = df.merge(adp, on="NAME", how="inner")
    globals()[position.lower()] = df
```

```python
qb.sort_values(by="ProxySharpe", ascending=False)
```
